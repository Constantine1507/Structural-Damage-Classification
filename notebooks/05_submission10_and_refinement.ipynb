# 05 — Submission 10 dan kandidat berikutnya

Public **0,60627** dilaporkan pengguna untuk submission 10. Notebook ini memeriksa CSV final dan membaca hasil 57 konfigurasi yang telah dijalankan. Sel notebook ini perlu dijalankan ulang; tidak melatih model. Lihat laporan untuk keterbatasan checkpoint.

```python
import json, hashlib
from pathlib import Path
import pandas as pd
from IPython.display import display
ROOT = Path.cwd() if Path('scripts').exists() else Path.cwd().parent
report = json.loads((ROOT/'reports/submission10_followup_results.json').read_text())
frames = {}
for key in ('submission10', 'candidate11'):
    info = report[key]
    path = ROOT/'submissions'/info['filename']
    assert hashlib.sha256(path.read_bytes()).hexdigest() == info['sha256']
    frame = pd.read_csv(path, dtype={'building_id':str})
    assert list(frame) == ['building_id','damage_grade']
    assert len(frame) == 228629 and frame.building_id.is_unique
    assert not frame.isna().any().any()
    assert frame.damage_grade.isin([1,2,3,4,5]).all()
    frames[key] = frame
assert frames['submission10'].building_id.equals(frames['candidate11'].building_id)
changed = (frames['submission10'].damage_grade != frames['candidate11'].damage_grade).sum()
assert changed == 685
print('CSV hashes and schema verified; changed rows:', changed)
```
